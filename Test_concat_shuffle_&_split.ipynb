```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import glob
import os
```

### Concatenar Excels test

```python
# Directorio donde guardar los Excels test PAC con los Abstracts
carpeta_Input_Excel = r"E:\PaperLLM\LLMzCor.github.io\Test\Human_labeled"
Test_file =r"E:\PaperLLM\LLMzCor.github.io\Test\test_file.xlsx"

def concatenate_xlsx(input_folder: str, output_file: str):
    excel_files = glob.glob(os.path.join(input_folder, "*.xlsx"))
    
    if not excel_files:
        print("No CSV files found in the specified directory.")
        return

    df_list = []
    for file in excel_files:
        df = pd.read_excel(file)
        df_list.append(df)

    concatenated_df = pd.concat(df_list, ignore_index=True)
    concatenated_df.to_excel(output_file, index=False)
    print(f"Concatenated {len(excel_files)} files into '{output_file}'.")

if __name__ == "__main__":
    input_folder = carpeta_Input_Excel
    output_file = Test_file 
    concatenate_xlsx(input_folder, output_file)
```

### Shuffle 

```python
# Ruta del archivo original y carpeta de salida
input_file = r"E:\PaperLLM\LLMzCor.github.io\Test\test_file.xlsx"
output_folder = r"E:\PaperLLM\LLMzCor.github.io\Test\Test_files"
os.makedirs(output_folder, exist_ok=True)
```

```python
# Leer el archivo original
df = pd.read_excel(input_file)
```

### Mezcla y reparte CSV 📚📝🎲

```python
# Mezclar las filas antes de dividir (para que la evaluación sea aleatoria)
df_shuffled = df.sample(frac=1, random_state=42).reset_index(drop=True)

# Calcular el tamaño del set de evaluación (30%)
eval_size = int(len(df_shuffled) * 0.3)

# Dividir en evaluación y entrenamiento
df_eval = df_shuffled.iloc[:eval_size]
df_train = df_shuffled.iloc[eval_size:]

# Guardar el set de evaluación como CSV
eval_path = os.path.join(output_folder, "Test_file_evaluation.csv")
df_eval.to_csv(eval_path, index=False, encoding="utf-8-sig")

# Crear 5 archivos de entrenamiento con el 70% de los datos, cada uno con distinto orden
for i, suffix in enumerate(["A", "B", "C", "D", "E"], 1):
    df_train_shuffled = df_train.sample(frac=1, random_state=i).reset_index(drop=True)
    train_path = os.path.join(output_folder, f"test_file_{suffix}.csv")
    df_train_shuffled.to_csv(train_path, index=False, encoding="utf-8-sig")

```
